```python
import torch
from diffusers import DiffusionPipeline

pipe = DiffusionPipeline.from_pretrained(
    "stabilityai/stable-diffusion-xl-base-1.0",
    torch_dtype=torch.float16,
    use_safetensors=True,
    variant="fp16",
)
pipe.to("cuda")
print("loaded, cuda:", torch.cuda.is_available())
```

```python
prompt = "warm watercolor storybook illustration, a Chinese grandmother and her grandchild sharing oranges at a wooden table by the window, soft morning light, cozy home, gentle colors, 1024x1024"
image = pipe(prompt=prompt, num_inference_steps=30, guidance_scale=7.0).images[0]
image.save("sdxl_test.png")
image
```
